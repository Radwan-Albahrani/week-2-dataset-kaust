## The big idea: what *is* a neural network?

Imagine teaching a young child to recognize the number **7**. You don't give them a rulebook —
you show them *thousands* of examples and correct them each time. Slowly, their brain figures out
the pattern on its own.

A **neural network** learns the exact same way:

1. We show it thousands of images, each labeled with the correct answer.
2. It makes a guess for each one.
3. We tell it how wrong it was.
4. It adjusts itself a tiny bit to be less wrong next time.
5. Repeat many times → it gets good!

Everything below is just how we do those 5 steps.

## Step 3 — Build the "brain"

Our brain is a stack of simple layers. Information flows through it top to bottom:

- **Flatten** — the image is a 32×32 color grid (3 color channels). We unroll it into one long
  line of `32 × 32 × 3 = 3072` numbers.
- **fc1 (first layer): 3072 → 128** — mixes the 3072 pixel-numbers into 128 "clues"
  the network invents (curves, edges, loops...).
- **ReLU** — a tiny rule: *"if a number is negative, set it to 0; otherwise keep it."*
  This on/off gate lets the network learn complicated patterns.
- **fc2 (last layer): 128 → 10** — turns the clues into **10 final scores**, one per category (0–9).
  The highest score is the network's guess.

## Step 4 — Set up the learning rules

Right now our brain is empty and guesses randomly. To help it improve we need two things:

- **The "grader" (loss)** — measures **how wrong** each guess was, as one number. **Big = very

- **The "coach" (optimizer)** — nudges the brain's internal knobs to reduce the mistake. The
  `lr=0.001` is how big each nudge is.

## Step 5 — Train the brain!

4. **Turn the knobs** — `optimizer.step()` makes the adjustment.
